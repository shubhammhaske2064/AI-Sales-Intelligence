```python
# INVESTIGATION: MISSING DISCOUNTS

missing_discount = sales_df[
    sales_df["Discount_Percent"].isna()
].copy()

print("MISSING DISCOUNT INVESTIGATION")
print(f"Records with missing discount: {len(missing_discount)}")


# 1. Basic statistics
print("NUMERICAL SUMMARY")

display(
    missing_discount[
        [
            "Quantity",
            "Unit_Price",
            "Sales_Amount",
            "Cost_Amount",
            "Profit"
        ]
    ].describe()
)


# 2. Check whether Sales_Amount equals Quantity × Unit_Price
missing_discount["Expected_Sales_No_Discount"] = (
    missing_discount["Quantity"] *
    missing_discount["Unit_Price"]
)

missing_discount["Sales_Difference"] = (
    missing_discount["Expected_Sales_No_Discount"]
    - missing_discount["Sales_Amount"]
)


print("DISCOUNT INFERENCE CHECK")
print(
    "Records where sales match full price:",
    (
        abs(missing_discount["Sales_Difference"]) < 0.01
    ).sum()
)

print(
    "Records where sales differ from full price:",
    (
        abs(missing_discount["Sales_Difference"]) >= 0.01
    ).sum()
)


# 3. Show examples
print("SAMPLE MISSING-DISCOUNT RECORDS")

display(
    missing_discount[
        [
            "Order_ID",
            "Quantity",
            "Unit_Price",
            "Sales_Amount",
            "Expected_Sales_No_Discount",
            "Sales_Difference",
            "Profit"
        ]
    ].head(20)
)


# 4. Estimate implied discount
missing_discount["Implied_Discount_Percent"] = (
    1 -
    missing_discount["Sales_Amount"] /
    missing_discount["Expected_Sales_No_Discount"]
) * 100

print("IMPLIED DISCOUNT")

display(
    missing_discount[
        "Implied_Discount_Percent"
    ].describe()
)
```

```python
# MISSING DISCOUNT - FINAL INVESTIGATION

missing_discount = sales_df[
    sales_df["Discount_Percent"].isna()
].copy()

# Calculate gross sales
missing_discount["Gross_Sales"] = (
    missing_discount["Quantity"] *
    missing_discount["Unit_Price"]
)

# Identify zero-quantity records
zero_quantity_missing_discount = missing_discount[
    missing_discount["Quantity"] == 0
]

print("ZERO-QUANTITY RECORDS AMONG MISSING DISCOUNTS")
print(
    "Zero-quantity records:",
    len(zero_quantity_missing_discount)
)


# Remove zero-quantity records for discount inference
valid_discount_records = missing_discount[
    missing_discount["Gross_Sales"] > 0
].copy()

# Calculate implied discount
valid_discount_records["Implied_Discount_Percent"] = (
    (
        valid_discount_records["Gross_Sales"]
        - valid_discount_records["Sales_Amount"]
    )
    / valid_discount_records["Gross_Sales"]
) * 100

valid_discount_records["Implied_Discount_Percent"] = (
    valid_discount_records["Implied_Discount_Percent"]
    .round(2)
)


print("IMPLIED DISCOUNT SUMMARY")
display(
    valid_discount_records[
        "Implied_Discount_Percent"
    ].describe()
)


# Categorize records
no_discount = valid_discount_records[
    valid_discount_records["Implied_Discount_Percent"].abs() < 0.01
]

discount_applied = valid_discount_records[
    valid_discount_records["Implied_Discount_Percent"] > 0.01
]


print("DISCOUNT CLASSIFICATION")
print("No discount / approximately 0%:",
      len(no_discount))

print("Discount appears to be applied:",
      len(discount_applied))

print("Zero quantity:",
      len(zero_quantity_missing_discount))


# Show implied discounts
print("\n" + "=" * 60)
print("SAMPLE IMPLIED DISCOUNTS")
display(
    valid_discount_records[
        [
            "Order_ID",
            "Quantity",
            "Unit_Price",
            "Gross_Sales",
            "Sales_Amount",
            "Implied_Discount_Percent"
        ]
    ].head(20)
)


# Check whether implied discounts fall within sensible range
print("IMPLIED DISCOUNT VALIDATION")
print(
    "Implied discount < 0%:",
    (
        valid_discount_records[
            "Implied_Discount_Percent"
        ] < 0
    ).sum()
)

print(
    "Implied discount > 100%:",
    (
        valid_discount_records[
            "Implied_Discount_Percent"
        ] > 100
    ).sum()
)
```

## Data Cleaning Decisions

### 1. Order_Date
- Convert from object to datetime.
- Invalid dates will be identified and handled.
- Current validation shows 0 invalid dates.

### 2. Missing Discount_Percent
- 19 records have implied discount ≈ 0%.
- 330 records have an identifiable implied discount.
- 1 record has Quantity = 0 and cannot be reliably inferred.
- For valid transactions, calculate:
  
  Implied Discount % =
  ((Quantity × Unit Price) - Sales Amount)
  / (Quantity × Unit Price) × 100

### 3. Channel
- Standardize inconsistent values such as `online` → `Online`.

### 4. Payment_Method
- Remove leading/trailing whitespace.
- Standardize values such as `UPI ` → `UPI`.

### 5. Quantity
- Quantity = 0 is considered an invalid transaction record.
- We will investigate before deciding whether to remove it.

### 6. Negative Profit
- Keep negative-profit transactions.
- Negative profit can represent legitimate loss-making sales.
- These records will be analyzed as a business insight.

### 7. Profit
- Profit calculation has been validated.
- No correction required because:
  
  Profit = Sales Amount - Cost Amount

### 8. Duplicate Records
- Complete duplicate rows and duplicate Order_IDs will be investigated.
- No records will be removed without evidence.

### 9. Order Status
- Completed, Cancelled and Pending transactions will be retained initially.
- Their treatment will depend on the analytical metric being calculated.

```python
# FACT SALES - DATA CLEANING PIPELINE

# Create a copy so the raw dataset remains untouched
sales_clean = sales_df.copy()

print("STARTING DATA CLEANING")

print(f"Original rows: {len(sales_clean):,}")


# ------------------------------------------------------------
# 1. Convert Order_Date to datetime
# ------------------------------------------------------------

sales_clean["Order_Date"] = pd.to_datetime(
    sales_clean["Order_Date"],
    errors="coerce"
)

print("\n1. Order_Date converted to datetime")


# ------------------------------------------------------------
# 2. Standardize text columns
# ------------------------------------------------------------

sales_clean["Channel"] = (
    sales_clean["Channel"]
    .astype("string")
    .str.strip()
    .str.title()
)

sales_clean["Payment_Method"] = (
    sales_clean["Payment_Method"]
    .astype("string")
    .str.strip()
)

print("2. Text fields standardized")


# ------------------------------------------------------------
# 3. Remove invalid quantity records
# ------------------------------------------------------------

invalid_quantity_mask = sales_clean["Quantity"] <= 0

invalid_quantity_count = invalid_quantity_mask.sum()

sales_clean = sales_clean[
    ~invalid_quantity_mask
].copy()

print(
    f"3. Removed invalid quantity records: "
    f"{invalid_quantity_count}"
)


# ------------------------------------------------------------
# 4. Reconstruct missing discounts
# ------------------------------------------------------------

missing_discount_mask = (
    sales_clean["Discount_Percent"].isna()
)

gross_sales = (
    sales_clean["Quantity"] *
    sales_clean["Unit_Price"]
)

# Calculate implied discount only where gross sales > 0
implied_discount = (
    (
        gross_sales -
        sales_clean["Sales_Amount"]
    )
    / gross_sales
) * 100

# Fill missing discounts using the transaction-level calculation
sales_clean.loc[
    missing_discount_mask,
    "Discount_Percent"
] = implied_discount[missing_discount_mask]

# Round discount
sales_clean["Discount_Percent"] = (
    sales_clean["Discount_Percent"]
    .round(2)
)

print(
    "4. Missing discounts reconstructed"
)


# ------------------------------------------------------------
# 5. Validate discount values
# ------------------------------------------------------------

invalid_discount = (
    (sales_clean["Discount_Percent"] < 0) |
    (sales_clean["Discount_Percent"] > 100)
)

print(
    "5. Invalid discount values after cleaning:",
    invalid_discount.sum()
)


# ------------------------------------------------------------
# 6. Validate profit calculation
# ------------------------------------------------------------

calculated_profit = (
    sales_clean["Sales_Amount"] -
    sales_clean["Cost_Amount"]
)

profit_difference = (
    sales_clean["Profit"] -
    calculated_profit
)

profit_errors = (
    abs(profit_difference) > 0.01
)

print(
    "6. Profit calculation errors:",
    profit_errors.sum()
)


# ------------------------------------------------------------
# 7. Save cleaned dataset
# ------------------------------------------------------------

output_path = (
    "C:/Users/shubh/Desktop/AI_Sales_Intelligence/data/cleaned/sales_clean.csv"
)

sales_clean.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 60)
print("CLEANING COMPLETE")
print("=" * 60)

print(f"Original rows : {len(sales_df):,}")
print(f"Cleaned rows  : {len(sales_clean):,}")
print(
    f"Rows removed  : "
    f"{len(sales_df) - len(sales_clean):,}"
)

print(f"\nSaved to: {output_path}")
```

```python
# ================================================================
# FIX DATE DIMENSION COVERAGE
# ================================================================

import pandas as pd

print("=" * 70)
print("FIXING DATE DIMENSION COVERAGE")
print("=" * 70)

# ------------------------------------------------
# 1. Identify required date range
# ------------------------------------------------

required_start = min(
    sales_model["Order_Date"].min(),
    return_model["Return_Date"].min()
)

required_end = max(
    sales_model["Order_Date"].max(),
    return_model["Return_Date"].max()
)

print(f"\nRequired date range:")
print(f"{required_start.date()} → {required_end.date()}")

print("\nCurrent date dimension:")
print(
    f"{date_model['Date'].min().date()} "
    f"→ "
    f"{date_model['Date'].max().date()}"
)

# ------------------------------------------------
# 2. Generate missing dates
# ------------------------------------------------

required_dates = pd.date_range(
    start=required_start,
    end=required_end,
    freq="D"
)

existing_dates = set(
    pd.to_datetime(date_model["Date"])
)

missing_dates = [
    d for d in required_dates
    if d not in existing_dates
]

print(f"\nMissing dates to add: {len(missing_dates)}")

# ------------------------------------------------
# 3. Add missing dates
# ------------------------------------------------

if missing_dates:

    new_dates = pd.DataFrame({
        "Date": missing_dates
    })

    # Create calendar attributes
    new_dates["Year"] = new_dates["Date"].dt.year
    new_dates["Month"] = new_dates["Date"].dt.month
    new_dates["Month_Name"] = new_dates["Date"].dt.month_name()
    new_dates["Quarter"] = (
        "Q" + new_dates["Date"].dt.quarter.astype(str)
    )

    new_dates["Day"] = new_dates["Date"].dt.day
    new_dates["Day_Name"] = new_dates["Date"].dt.day_name()
    new_dates["Week_Number"] = (
        new_dates["Date"].dt.isocalendar().week.astype(int)
    )

    new_dates["Is_Weekend"] = (
        new_dates["Date"].dt.dayofweek >= 5
    )

    # Match the existing date_model column structure
    new_dates = new_dates[
        [col for col in date_model.columns if col in new_dates.columns]
    ]

    # Add any columns that exist in date_model
    # but were not generated above
    for col in date_model.columns:
        if col not in new_dates.columns:
            new_dates[col] = pd.NA

    new_dates = new_dates[date_model.columns]

    # Combine
    date_model = pd.concat(
        [date_model, new_dates],
        ignore_index=True
    )

    # Remove accidental duplicates
    date_model = (
        date_model
        .drop_duplicates(subset=["Date"])
        .sort_values("Date")
        .reset_index(drop=True)
    )

# ------------------------------------------------
# 4. Display result
# ------------------------------------------------

print("\nUpdated date dimension:")
print(f"Rows: {len(date_model):,}")
print(
    f"Date range: "
    f"{date_model['Date'].min().date()} "
    f"→ "
    f"{date_model['Date'].max().date()}"
)

print("\nNewly added dates:")
display(
    date_model[
        date_model["Date"].isin(missing_dates)
    ].head(20)
)

# ------------------------------------------------
# 5. Validate date uniqueness
# ------------------------------------------------

duplicate_dates = date_model["Date"].duplicated().sum()

print("\n" + "=" * 70)
print("DATE DIMENSION VALIDATION")
print("=" * 70)

print(f"Duplicate dates : {duplicate_dates}")
print(f"Missing dates   : {date_model['Date'].isna().sum()}")

# ------------------------------------------------
# 6. Validate Sales → Date
# ------------------------------------------------

sales_date_missing = (
    ~pd.to_datetime(sales_model["Order_Date"])
    .dt.normalize()
    .isin(pd.to_datetime(date_model["Date"]))
).sum()

# ------------------------------------------------
# 7. Validate Return → Date
# ------------------------------------------------

return_date_missing = (
    ~pd.to_datetime(return_model["Return_Date"])
    .dt.normalize()
    .isin(pd.to_datetime(date_model["Date"]))
).sum()

print("\nRelationship validation:")
print(
    f"Sales → Date   : "
    f"{'PASS' if sales_date_missing == 0 else 'FAIL'}"
)

print(
    f"Return → Date  : "
    f"{'PASS' if return_date_missing == 0 else 'FAIL'}"
)

print(f"\nUnmatched Sales dates  : {sales_date_missing}")
print(f"Unmatched Return dates : {return_date_missing}")

# ------------------------------------------------
# 8. Save updated date model
# ------------------------------------------------

date_model_path = cleaned_path / "date_model.csv"

date_model.to_csv(
    date_model_path,
    index=False
)

print("\nSaved:")
print(date_model_path)

print("\n" + "=" * 70)
print("DATE DIMENSION FIX COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SQL KPI 1 — OVERALL SALES PERFORMANCE
# ============================================================

sales_kpi_query = """
SELECT
    SUM(Sales_Amount) AS Total_Revenue,
    SUM(Cost_Amount) AS Total_Cost,
    SUM(Profit) AS Total_Profit,
    COUNT(DISTINCT Order_ID) AS Total_Orders,
    SUM(Quantity) AS Units_Sold,

    ROUND(
        SUM(Profit) * 100.0 / NULLIF(SUM(Sales_Amount), 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(
        SUM(Sales_Amount) * 1.0 /
        NULLIF(COUNT(DISTINCT Order_ID), 0),
        2
    ) AS Average_Order_Value,

    ROUND(
        SUM(Sales_Amount) * 1.0 /
        NULLIF(SUM(Quantity), 0),
        2
    ) AS Average_Selling_Price

FROM sales;
"""

sales_kpi = pd.read_sql_query(sales_kpi_query, conn)

display(sales_kpi)
```

```python
# ============================================================
# SQL KPI 2 — MONTHLY SALES PERFORMANCE
# ============================================================

monthly_sales_query = """
SELECT
    Year,
    Month,
    Month_Name,

    SUM(Sales_Amount) AS Revenue,
    SUM(Cost_Amount) AS Cost,
    SUM(Profit) AS Profit,
    SUM(Quantity) AS Units_Sold,
    COUNT(DISTINCT Order_ID) AS Orders,

    ROUND(
        SUM(Profit) * 100.0 /
        NULLIF(SUM(Sales_Amount), 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(
        SUM(Sales_Amount) * 1.0 /
        NULLIF(COUNT(DISTINCT Order_ID), 0),
        2
    ) AS Average_Order_Value

FROM sales

GROUP BY
    Year,
    Month,
    Month_Name

ORDER BY
    Year,
    Month;
"""

monthly_sales = pd.read_sql_query(monthly_sales_query, conn)

display(monthly_sales.head(12))
```

```python
# ============================================================
# SQL KPI 3 — CUSTOMER PERFORMANCE
# ============================================================

customer_performance_query = """
SELECT
    s.Customer_ID,

    c.Customer_Name,

    COUNT(DISTINCT s.Order_ID) AS Total_Orders,

    SUM(s.Quantity) AS Units_Sold,

    ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

    ROUND(SUM(s.Profit), 2) AS Profit,

    ROUND(
        SUM(s.Profit) * 100.0 /
        NULLIF(SUM(s.Sales_Amount), 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(
        SUM(s.Sales_Amount) * 1.0 /
        NULLIF(COUNT(DISTINCT s.Order_ID), 0),
        2
    ) AS Average_Order_Value

FROM sales s

LEFT JOIN customer c
    ON s.Customer_ID = c.Customer_ID

GROUP BY
    s.Customer_ID,
    c.Customer_Name

ORDER BY
    Revenue DESC;
"""

customer_performance = pd.read_sql_query(
    customer_performance_query,
    conn
)

display(customer_performance.head(20))
```

```python
# ============================================================
# SQL KPI 4 — PRODUCT PERFORMANCE
# ============================================================

product_performance_query = """
SELECT
    s.Product_ID,

    p.Product_Name,
    p.Category,

    SUM(s.Quantity) AS Units_Sold,

    ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

    ROUND(SUM(s.Profit), 2) AS Profit,

    ROUND(
        SUM(s.Profit) * 100.0 /
        NULLIF(SUM(s.Sales_Amount), 0),
        2
    ) AS Profit_Margin_Percent,

    COUNT(DISTINCT s.Order_ID) AS Orders

FROM sales s

LEFT JOIN product p
    ON s.Product_ID = p.Product_ID

GROUP BY
    s.Product_ID,
    p.Product_Name,
    p.Category

ORDER BY
    Revenue DESC;
"""

product_performance = pd.read_sql_query(
    product_performance_query,
    conn
)

display(product_performance.head(20))
```

```python
# ============================================================
# SQL KPI 5 — REGIONAL PERFORMANCE
# ============================================================

regional_performance_query = """
SELECT
    s.Region_ID,

    r.Region,

    COUNT(DISTINCT s.Order_ID) AS Orders,

    COUNT(DISTINCT s.Customer_ID) AS Customers,

    SUM(s.Quantity) AS Units_Sold,

    ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

    ROUND(SUM(s.Profit), 2) AS Profit,

    ROUND(
        SUM(s.Profit) * 100.0 /
        NULLIF(SUM(s.Sales_Amount), 0),
        2
    ) AS Profit_Margin_Percent

FROM sales s

LEFT JOIN region r
    ON s.Region_ID = r.Region_ID

GROUP BY
    s.Region_ID,
    r.Region

ORDER BY
    Revenue DESC;
"""

regional_performance = pd.read_sql_query(
    regional_performance_query,
    conn
)

display(regional_performance)
```

```python
# ============================================================
# SQL KPI 6 — DISCOUNT VS PROFITABILITY
# ============================================================

discount_analysis_query = """
SELECT

    CASE
        WHEN Discount_Percent = 0 THEN '0%'
        WHEN Discount_Percent <= 5 THEN '1-5%'
        WHEN Discount_Percent <= 10 THEN '6-10%'
        WHEN Discount_Percent <= 15 THEN '11-15%'
        WHEN Discount_Percent <= 20 THEN '16-20%'
        ELSE '20%+'
    END AS Discount_Band,

    COUNT(*) AS Transactions,

    ROUND(SUM(Sales_Amount), 2) AS Revenue,

    ROUND(SUM(Profit), 2) AS Profit,

    ROUND(
        SUM(Profit) * 100.0 /
        NULLIF(SUM(Sales_Amount), 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(AVG(Discount_Percent), 2) AS Average_Discount

FROM sales

GROUP BY Discount_Band

ORDER BY
    MIN(Discount_Percent);
"""

discount_analysis = pd.read_sql_query(
    discount_analysis_query,
    conn
)

display(discount_analysis)
```

```python
# ============================================================
# SQL KPI 7 — VALID RETURN ANALYSIS
# ============================================================

return_analysis_query = """
SELECT

    COUNT(*) AS Valid_Return_Records,

    SUM(Analytical_Return_Quantity) AS Valid_Return_Quantity,

    ROUND(
        SUM(Analytical_Refund_Amount),
        2
    ) AS Refund_Amount,

    ROUND(
        SUM(Analytical_Refund_Amount) /
        NULLIF(COUNT(*), 0),
        2
    ) AS Average_Refund_Amount

FROM return

WHERE Is_Valid_Return = 1;
"""

return_analysis = pd.read_sql_query(
    return_analysis_query,
    conn
)

display(return_analysis)
```

```python
# ============================================================
# SQL KPI 8 — RETURN RATE
# ============================================================

return_rate_query = """
WITH sales_units AS (

    SELECT
        SUM(Quantity) AS Units_Sold
    FROM sales

),

valid_returns AS (

    SELECT
        SUM(Analytical_Return_Quantity) AS Returned_Units
    FROM return
    WHERE Is_Valid_Return = 1

)

SELECT

    sales_units.Units_Sold,

    valid_returns.Returned_Units,

    ROUND(
        valid_returns.Returned_Units * 100.0 /
        NULLIF(sales_units.Units_Sold, 0),
        2
    ) AS Return_Rate_Percent

FROM sales_units
CROSS JOIN valid_returns;
"""

return_rate = pd.read_sql_query(
    return_rate_query,
    conn
)

display(return_rate)
```

```python
# ============================================
# ADVANCED SQL 1: MoM & YoY Revenue Growth
# Using LAG() Window Function
# ============================================

mom_yoy_query = """
WITH monthly_revenue AS (
    SELECT
        d.Year,
        d.Month_Number,
        d.Month_Name,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue

    FROM sales s

    LEFT JOIN date d
        ON DATE(s.Order_Date) = DATE(d.Date)

    GROUP BY
        d.Year,
        d.Month_Number,
        d.Month_Name
),

revenue_with_previous AS (
    SELECT
        Year,
        Month_Number,
        Month_Name,
        Revenue,

        -- Previous month's revenue
        LAG(Revenue) OVER (
            ORDER BY Year, Month_Number
        ) AS Previous_Month_Revenue,

        -- Same month in previous year
        LAG(Revenue, 12) OVER (
            ORDER BY Year, Month_Number
        ) AS Previous_Year_Revenue

    FROM monthly_revenue
)

SELECT
    Year,
    Month_Number,
    Month_Name,
    Revenue,

    ROUND(Previous_Month_Revenue, 2)
        AS Previous_Month_Revenue,

    ROUND(
        (Revenue - Previous_Month_Revenue) * 100.0
        / NULLIF(Previous_Month_Revenue, 0),
        2
    ) AS MoM_Growth_Percent,

    ROUND(Previous_Year_Revenue, 2)
        AS Previous_Year_Revenue,

    ROUND(
        (Revenue - Previous_Year_Revenue) * 100.0
        / NULLIF(Previous_Year_Revenue, 0),
        2
    ) AS YoY_Growth_Percent

FROM revenue_with_previous

ORDER BY
    Year,
    Month_Number;
"""

monthly_growth = pd.read_sql_query(
    mom_yoy_query,
    conn
)

display(monthly_growth)
```

```python
# ============================================
# ADVANCED SQL 2: Product Ranking
# Using RANK() and ROW_NUMBER()
# ============================================

product_ranking_query = """
WITH product_performance AS (
    SELECT
        p.Product_ID,
        p.Product_Name,
        p.Category,

        SUM(s.Quantity) AS Units_Sold,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Profit), 2) AS Profit,

        ROUND(
            SUM(s.Profit) * 100.0 /
            NULLIF(SUM(s.Sales_Amount), 0),
            2
        ) AS Profit_Margin_Percent,

        COUNT(DISTINCT s.Order_ID) AS Orders

    FROM sales s

    LEFT JOIN product p
        ON s.Product_ID = p.Product_ID

    GROUP BY
        p.Product_ID,
        p.Product_Name,
        p.Category
),

ranked_products AS (
    SELECT
        Product_ID,
        Product_Name,
        Category,
        Units_Sold,
        Revenue,
        Profit,
        Profit_Margin_Percent,
        Orders,

        -- Ranking based on revenue
        RANK() OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Rank,

        -- Ranking based on profit
        RANK() OVER (
            ORDER BY Profit DESC
        ) AS Profit_Rank,

        -- Unique sequential ranking
        ROW_NUMBER() OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Row_Number

    FROM product_performance
)

SELECT *
FROM ranked_products
ORDER BY Revenue_Rank;
"""

product_ranking = pd.read_sql_query(
    product_ranking_query,
    conn
)

display(product_ranking)
```

```python
# ============================================
# ADVANCED SQL 3: Customer Ranking
# Using RANK(), ROW_NUMBER() and NTILE()
# ============================================

customer_ranking_query = """
WITH customer_performance AS (
    SELECT
        c.Customer_ID,
        c.Customer_Name,

        COUNT(DISTINCT s.Order_ID) AS Total_Orders,

        SUM(s.Quantity) AS Units_Sold,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Profit), 2) AS Profit,

        ROUND(
            SUM(s.Profit) * 100.0 /
            NULLIF(SUM(s.Sales_Amount), 0),
            2
        ) AS Profit_Margin_Percent

    FROM sales s

    LEFT JOIN customer c
        ON s.Customer_ID = c.Customer_ID

    GROUP BY
        c.Customer_ID,
        c.Customer_Name
),

ranked_customers AS (
    SELECT
        Customer_ID,
        Customer_Name,
        Total_Orders,
        Units_Sold,
        Revenue,
        Profit,
        Profit_Margin_Percent,

        -- Revenue ranking
        RANK() OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Rank,

        -- Profit ranking
        RANK() OVER (
            ORDER BY Profit DESC
        ) AS Profit_Rank,

        -- Unique customer ordering
        ROW_NUMBER() OVER (
            ORDER BY Revenue DESC
        ) AS Customer_Row_Number,

        -- Divide customers into 4 groups based on revenue
        NTILE(4) OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Quartile

    FROM customer_performance
)

SELECT
    *,
    CASE
        WHEN Revenue_Quartile = 1 THEN 'High Value'
        WHEN Revenue_Quartile = 2 THEN 'Medium-High Value'
        WHEN Revenue_Quartile = 3 THEN 'Medium-Low Value'
        WHEN Revenue_Quartile = 4 THEN 'Low Value'
    END AS Customer_Value_Segment

FROM ranked_customers

ORDER BY Revenue_Rank;
"""

customer_ranking = pd.read_sql_query(
    customer_ranking_query,
    conn
)

display(customer_ranking)
```

```python
# ============================================
# ADVANCED SQL 4: Target vs Actual Performance
# ============================================

target_vs_actual_query = """
WITH actual_sales AS (
    SELECT
        s.Region_ID,
        d.Year,
        d.Month_Number,

        ROUND(SUM(s.Sales_Amount), 2) AS Actual_Revenue,
        ROUND(SUM(s.Profit), 2) AS Actual_Profit,
        SUM(s.Quantity) AS Actual_Units,
        COUNT(DISTINCT s.Order_ID) AS Actual_Orders

    FROM sales s

    LEFT JOIN date d
        ON DATE(s.Order_Date) = DATE(d.Date)

    GROUP BY
        s.Region_ID,
        d.Year,
        d.Month_Number
),

target_data AS (
    SELECT
        t.Region_ID,
        d.Year,
        d.Month_Number,

        ROUND(SUM(t.Sales_Target), 2) AS Target_Revenue,
        ROUND(SUM(t.Profit_Target), 2) AS Target_Profit

    FROM target t

    LEFT JOIN date d
        ON DATE(t.Date) = DATE(d.Date)

    GROUP BY
        t.Region_ID,
        d.Year,
        d.Month_Number
)

SELECT
    a.Region_ID,
    r.Region,

    a.Year,
    a.Month_Number,

    a.Actual_Revenue,
    t.Target_Revenue,

    ROUND(
        a.Actual_Revenue - t.Target_Revenue,
        2
    ) AS Revenue_Variance,

    ROUND(
        a.Actual_Revenue * 100.0 /
        NULLIF(t.Target_Revenue, 0),
        2
    ) AS Revenue_Achievement_Percent,

    a.Actual_Profit,
    t.Target_Profit,

    ROUND(
        a.Actual_Profit - t.Target_Profit,
        2
    ) AS Profit_Variance,

    ROUND(
        a.Actual_Profit * 100.0 /
        NULLIF(t.Target_Profit, 0),
        2
    ) AS Profit_Achievement_Percent,

    a.Actual_Units,
    a.Actual_Orders

FROM actual_sales a

LEFT JOIN target_data t
    ON a.Region_ID = t.Region_ID
    AND a.Year = t.Year
    AND a.Month_Number = t.Month_Number

LEFT JOIN region r
    ON a.Region_ID = r.Region_ID

ORDER BY
    a.Year,
    a.Month_Number,
    a.Region_ID;
"""

target_vs_actual = pd.read_sql_query(
    target_vs_actual_query,
    conn
)

display(target_vs_actual)
```

```python
# ============================================
# ADVANCED SQL 5: Regional Ranking
# ============================================

regional_ranking_query = """
WITH regional_sales AS (
    SELECT
        s.Region_ID,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Profit), 2) AS Profit,

        SUM(s.Quantity) AS Units_Sold,

        COUNT(DISTINCT s.Order_ID) AS Orders,

        COUNT(DISTINCT s.Customer_ID) AS Customers,

        ROUND(
            SUM(s.Profit) * 100.0 /
            NULLIF(SUM(s.Sales_Amount), 0),
            2
        ) AS Profit_Margin_Percent

    FROM sales s

    GROUP BY
        s.Region_ID
),

regional_targets AS (
    SELECT
        t.Region_ID,

        ROUND(SUM(t.Sales_Target), 2) AS Target_Revenue,

        ROUND(SUM(t.Profit_Target), 2) AS Target_Profit

    FROM target t

    GROUP BY
        t.Region_ID
),

regional_analysis AS (
    SELECT
        rs.Region_ID,
        r.Region,

        rs.Revenue,
        rt.Target_Revenue,

        ROUND(
            rs.Revenue - rt.Target_Revenue,
            2
        ) AS Revenue_Variance,

        ROUND(
            rs.Revenue * 100.0 /
            NULLIF(rt.Target_Revenue, 0),
            2
        ) AS Revenue_Achievement_Percent,

        rs.Profit,
        rt.Target_Profit,

        ROUND(
            rs.Profit - rt.Target_Profit,
            2
        ) AS Profit_Variance,

        ROUND(
            rs.Profit * 100.0 /
            NULLIF(rt.Target_Profit, 0),
            2
        ) AS Profit_Achievement_Percent,

        rs.Profit_Margin_Percent,
        rs.Units_Sold,
        rs.Orders,
        rs.Customers

    FROM regional_sales rs

    LEFT JOIN regional_targets rt
        ON rs.Region_ID = rt.Region_ID

    LEFT JOIN region r
        ON rs.Region_ID = r.Region_ID
)

SELECT
    *,

    RANK() OVER (
        ORDER BY Revenue DESC
    ) AS Revenue_Rank,

    RANK() OVER (
        ORDER BY Profit DESC
    ) AS Profit_Rank,

    RANK() OVER (
        ORDER BY Profit_Margin_Percent DESC
    ) AS Margin_Rank,

    RANK() OVER (
        ORDER BY Revenue_Achievement_Percent DESC
    ) AS Target_Achievement_Rank

FROM regional_analysis

ORDER BY Revenue_Rank;
"""

regional_ranking = pd.read_sql_query(
    regional_ranking_query,
    conn
)

display(regional_ranking)
```

```python
# ============================================
# ADVANCED SQL 6: TOP & BOTTOM PRODUCT ANALYSIS
# ============================================

top_bottom_products_query = """
WITH product_performance AS (
    SELECT
        p.Product_ID,
        p.Product_Name,
        p.Category,

        SUM(s.Quantity) AS Units_Sold,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Profit), 2) AS Profit,

        ROUND(
            SUM(s.Profit) * 100.0 /
            NULLIF(SUM(s.Sales_Amount), 0),
            2
        ) AS Profit_Margin_Percent,

        COUNT(DISTINCT s.Order_ID) AS Orders

    FROM sales s

    LEFT JOIN product p
        ON s.Product_ID = p.Product_ID

    GROUP BY
        p.Product_ID,
        p.Product_Name,
        p.Category
),

ranked_products AS (
    SELECT
        *,

        RANK() OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Rank,

        RANK() OVER (
            ORDER BY Profit DESC
        ) AS Profit_Rank,

        RANK() OVER (
            ORDER BY Profit_Margin_Percent DESC
        ) AS Margin_Rank

    FROM product_performance
),

benchmarks AS (
    SELECT
        AVG(Revenue) AS Average_Revenue,
        AVG(Profit_Margin_Percent) AS Average_Margin
    FROM product_performance
)

SELECT
    rp.Product_ID,
    rp.Product_Name,
    rp.Category,
    rp.Units_Sold,
    rp.Orders,
    rp.Revenue,
    rp.Profit,
    rp.Profit_Margin_Percent,

    rp.Revenue_Rank,
    rp.Profit_Rank,
    rp.Margin_Rank,

    CASE
        WHEN rp.Revenue_Rank <= 10
            THEN 'Top 10 Revenue'

        WHEN rp.Revenue_Rank >= (
            SELECT MAX(Revenue_Rank)
            FROM ranked_products
        ) - 9
            THEN 'Bottom 10 Revenue'

        ELSE 'Middle'
    END AS Revenue_Performance,

    CASE
        WHEN rp.Revenue > b.Average_Revenue
             AND rp.Profit_Margin_Percent < b.Average_Margin
            THEN 'High Revenue - Low Margin'

        WHEN rp.Revenue < b.Average_Revenue
             AND rp.Profit_Margin_Percent > b.Average_Margin
            THEN 'Low Revenue - High Margin'

        WHEN rp.Revenue > b.Average_Revenue
             AND rp.Profit_Margin_Percent > b.Average_Margin
            THEN 'High Revenue - High Margin'

        ELSE 'Other'
    END AS Business_Performance_Type

FROM ranked_products rp

CROSS JOIN benchmarks b

ORDER BY
    rp.Revenue_Rank;
"""

top_bottom_products = pd.read_sql_query(
    top_bottom_products_query,
    conn
)

display(top_bottom_products)
```

```python
# ============================================
# ADVANCED SQL 7: CUSTOMER REVENUE CONCENTRATION
# ============================================

customer_concentration_query = """
WITH customer_sales AS (
    SELECT
        c.Customer_ID,
        c.Customer_Name,

        COUNT(DISTINCT s.Order_ID) AS Total_Orders,

        SUM(s.Quantity) AS Units_Sold,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Profit), 2) AS Profit

    FROM sales s

    LEFT JOIN customer c
        ON s.Customer_ID = c.Customer_ID

    GROUP BY
        c.Customer_ID,
        c.Customer_Name
),

customer_metrics AS (
    SELECT
        *,

        RANK() OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Rank,

        SUM(Revenue) OVER () AS Total_Revenue

    FROM customer_sales
),

concentration AS (
    SELECT
        *,

        ROUND(
            Revenue * 100.0 /
            NULLIF(Total_Revenue, 0),
            4
        ) AS Revenue_Contribution_Percent,

        ROUND(
            SUM(Revenue) OVER (
                ORDER BY Revenue DESC
                ROWS BETWEEN UNBOUNDED PRECEDING
                AND CURRENT ROW
            ) * 100.0 /
            NULLIF(Total_Revenue, 0),
            4
        ) AS Cumulative_Revenue_Percent

    FROM customer_metrics
)

SELECT
    Customer_ID,
    Customer_Name,
    Total_Orders,
    Units_Sold,
    Revenue,
    Profit,
    Revenue_Rank,
    Revenue_Contribution_Percent,
    Cumulative_Revenue_Percent,

    CASE
        WHEN Cumulative_Revenue_Percent <= 20
            THEN 'Top Revenue Contributors'

        WHEN Cumulative_Revenue_Percent <= 50
            THEN 'Core Revenue Contributors'

        WHEN Cumulative_Revenue_Percent <= 80
            THEN 'Moderate Contributors'

        ELSE 'Long Tail Customers'
    END AS Revenue_Concentration_Segment

FROM concentration

ORDER BY
    Revenue_Rank;
"""

customer_concentration = pd.read_sql_query(
    customer_concentration_query,
    conn
)

display(customer_concentration)
```

```python
# ============================================
# ADVANCED SQL 8: DISCOUNT EFFECTIVENESS
# ============================================

discount_effectiveness_query = """
WITH discount_analysis AS (
    SELECT
        CASE
            WHEN Discount_Percent = 0 THEN '0% Discount'
            WHEN Discount_Percent > 0
                 AND Discount_Percent <= 5 THEN '1-5% Discount'
            WHEN Discount_Percent > 5
                 AND Discount_Percent <= 10 THEN '6-10% Discount'
            WHEN Discount_Percent > 10
                 AND Discount_Percent <= 15 THEN '11-15% Discount'
            WHEN Discount_Percent > 15
                 AND Discount_Percent <= 20 THEN '16-20% Discount'
            ELSE '20%+ Discount'
        END AS Discount_Band,

        COUNT(DISTINCT Order_ID) AS Orders,

        SUM(Quantity) AS Units_Sold,

        ROUND(SUM(Sales_Amount), 2) AS Revenue,

        ROUND(SUM(Profit), 2) AS Profit,

        ROUND(AVG(Discount_Percent), 2) AS Average_Discount

    FROM sales

    GROUP BY
        CASE
            WHEN Discount_Percent = 0 THEN '0% Discount'
            WHEN Discount_Percent > 0
                 AND Discount_Percent <= 5 THEN '1-5% Discount'
            WHEN Discount_Percent > 5
                 AND Discount_Percent <= 10 THEN '6-10% Discount'
            WHEN Discount_Percent > 10
                 AND Discount_Percent <= 15 THEN '11-15% Discount'
            WHEN Discount_Percent > 15
                 AND Discount_Percent <= 20 THEN '16-20% Discount'
            ELSE '20%+ Discount'
        END
),

discount_metrics AS (
    SELECT
        *,

        ROUND(
            Profit * 100.0 /
            NULLIF(Revenue, 0),
            2
        ) AS Profit_Margin_Percent,

        ROUND(
            Revenue / NULLIF(Orders, 0),
            2
        ) AS Average_Order_Value

    FROM discount_analysis
)

SELECT
    Discount_Band,
    Orders,
    Units_Sold,
    Revenue,
    Profit,
    Profit_Margin_Percent,
    Average_Discount,
    Average_Order_Value

FROM discount_metrics

ORDER BY
    CASE Discount_Band
        WHEN '0% Discount' THEN 1
        WHEN '1-5% Discount' THEN 2
        WHEN '6-10% Discount' THEN 3
        WHEN '11-15% Discount' THEN 4
        WHEN '16-20% Discount' THEN 5
        WHEN '20%+ Discount' THEN 6
    END;
"""

discount_effectiveness = pd.read_sql_query(
    discount_effectiveness_query,
    conn
)

display(discount_effectiveness)
```

```python
discount_effectiveness["Revenue_Contribution_Percent"] = (
    discount_effectiveness["Revenue"]
    / discount_effectiveness["Revenue"].sum()
    * 100
)

discount_effectiveness["Profit_Contribution_Percent"] = (
    discount_effectiveness["Profit"]
    / discount_effectiveness["Profit"].sum()
    * 100
)

discount_effectiveness = discount_effectiveness.round(2)

display(discount_effectiveness)
```

```python
# ============================================
# ADVANCED SQL 9: RETURN & PRODUCT IMPACT
# ============================================

return_product_query = """
WITH product_sales AS (
    SELECT
        p.Product_ID,
        p.Product_Name,
        p.Category,

        SUM(s.Quantity) AS Units_Sold,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Profit), 2) AS Profit

    FROM sales s

    LEFT JOIN product p
        ON s.Product_ID = p.Product_ID

    GROUP BY
        p.Product_ID,
        p.Product_Name,
        p.Category
),

product_returns AS (
    SELECT
        Product_ID,

        SUM(
            CASE
                WHEN Is_Valid_Return = 1
                THEN Analytical_Return_Quantity
                ELSE 0
            END
        ) AS Returned_Units,

        ROUND(
            SUM(
                CASE
                    WHEN Is_Valid_Return = 1
                    THEN Analytical_Refund_Amount
                    ELSE 0
                END
            ),
            2
        ) AS Refund_Amount,

        COUNT(
            CASE
                WHEN Is_Valid_Return = 1
                THEN Return_ID
            END
        ) AS Valid_Return_Records

    FROM return

    GROUP BY
        Product_ID
)

SELECT
    ps.Product_ID,
    ps.Product_Name,
    ps.Category,

    ps.Units_Sold,
    ps.Revenue,
    ps.Profit,

    COALESCE(pr.Returned_Units, 0) AS Returned_Units,

    COALESCE(pr.Refund_Amount, 0) AS Refund_Amount,

    COALESCE(pr.Valid_Return_Records, 0)
        AS Valid_Return_Records,

    ROUND(
        COALESCE(pr.Returned_Units, 0) * 100.0 /
        NULLIF(ps.Units_Sold, 0),
        2
    ) AS Return_Rate_Percent,

    ROUND(
        COALESCE(pr.Refund_Amount, 0) * 100.0 /
        NULLIF(ps.Revenue, 0),
        2
    ) AS Refund_to_Revenue_Percent,

    ROUND(
        ps.Profit * 100.0 /
        NULLIF(ps.Revenue, 0),
        2
    ) AS Profit_Margin_Percent

FROM product_sales ps

LEFT JOIN product_returns pr
    ON ps.Product_ID = pr.Product_ID

ORDER BY
    Return_Rate_Percent DESC;
"""

return_product_analysis = pd.read_sql_query(
    return_product_query,
    conn
)

display(return_product_analysis)
```

```python
overall_return_rate = (
    return_product_analysis["Returned_Units"].sum()
    /
    return_product_analysis["Units_Sold"].sum()
    * 100
)

return_product_analysis["Return_Risk"] = (
    return_product_analysis["Return_Rate_Percent"]
    .apply(
        lambda x:
        "High Return Risk"
        if x > overall_return_rate * 1.5
        else
        "Moderate Return Risk"
        if x > overall_return_rate
        else
        "Low Return Risk"
    )
)

display(
    return_product_analysis[
        [
            "Product_ID",
            "Product_Name",
            "Category",
            "Units_Sold",
            "Returned_Units",
            "Return_Rate_Percent",
            "Refund_Amount",
            "Return_Risk"
        ]
    ].sort_values(
        "Return_Rate_Percent",
        ascending=False
    )
)
```

```python
# ============================================
# ADVANCED SQL 10: MONTHLY BUSINESS PERFORMANCE
# ============================================

monthly_business_query = """
WITH monthly_performance AS (
    SELECT
        d.Year,
        d.Month_Number,
        d.Month_Name,

        SUM(s.Quantity) AS Units_Sold,

        COUNT(DISTINCT s.Order_ID) AS Orders,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Cost_Amount), 2) AS Cost,

        ROUND(SUM(s.Profit), 2) AS Profit

    FROM sales s

    LEFT JOIN date d
        ON DATE(s.Order_Date) = DATE(d.Date)

    GROUP BY
        d.Year,
        d.Month_Number,
        d.Month_Name
),

monthly_growth AS (
    SELECT
        *,

        LAG(Revenue) OVER (
            ORDER BY Year, Month_Number
        ) AS Previous_Month_Revenue,

        LAG(Profit) OVER (
            ORDER BY Year, Month_Number
        ) AS Previous_Month_Profit

    FROM monthly_performance
)

SELECT
    Year,
    Month_Number,
    Month_Name,

    Units_Sold,
    Orders,

    Revenue,
    Cost,
    Profit,

    ROUND(
        Profit * 100.0 /
        NULLIF(Revenue, 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(
        Revenue / NULLIF(Orders, 0),
        2
    ) AS Average_Order_Value,

    ROUND(Previous_Month_Revenue, 2)
        AS Previous_Month_Revenue,

    ROUND(
        (Revenue - Previous_Month_Revenue) * 100.0 /
        NULLIF(Previous_Month_Revenue, 0),
        2
    ) AS MoM_Revenue_Growth_Percent,

    ROUND(Previous_Month_Profit, 2)
        AS Previous_Month_Profit,

    ROUND(
        (Profit - Previous_Month_Profit) * 100.0 /
        NULLIF(Previous_Month_Profit, 0),
        2
    ) AS MoM_Profit_Growth_Percent

FROM monthly_growth

ORDER BY
    Year,
    Month_Number;
"""

monthly_business = pd.read_sql_query(
    monthly_business_query,
    conn
)

display(monthly_business)
```

```python
# ============================================
# ADVANCED SQL 11: CUSTOMER RFM SEGMENTATION
# ============================================

rfm_query = """
WITH customer_metrics AS (
    SELECT
        c.Customer_ID,
        c.Customer_Name,

        MAX(DATE(s.Order_Date)) AS Last_Purchase_Date,

        COUNT(DISTINCT s.Order_ID) AS Frequency,

        ROUND(
            SUM(s.Sales_Amount),
            2
        ) AS Monetary

    FROM sales s

    LEFT JOIN customer c
        ON s.Customer_ID = c.Customer_ID

    GROUP BY
        c.Customer_ID,
        c.Customer_Name
),

reference_date AS (
    SELECT
        MAX(DATE(Order_Date)) AS Analysis_Date
    FROM sales
),

rfm_base AS (
    SELECT
        cm.Customer_ID,
        cm.Customer_Name,

        julianday(rd.Analysis_Date)
        - julianday(cm.Last_Purchase_Date)
        AS Recency,

        cm.Frequency,
        cm.Monetary

    FROM customer_metrics cm

    CROSS JOIN reference_date rd
),

rfm_scores AS (
    SELECT
        *,

        NTILE(5) OVER (
            ORDER BY Recency DESC
        ) AS Recency_Score,

        NTILE(5) OVER (
            ORDER BY Frequency ASC
        ) AS Frequency_Score,

        NTILE(5) OVER (
            ORDER BY Monetary ASC
        ) AS Monetary_Score

    FROM rfm_base
)

SELECT
    Customer_ID,
    Customer_Name,

    ROUND(Recency, 0) AS Recency_Days,

    Frequency,
    Monetary,

    Recency_Score,
    Frequency_Score,
    Monetary_Score,

    (
        Recency_Score
        + Frequency_Score
        + Monetary_Score
    ) AS RFM_Total_Score,

    CASE

        WHEN Recency_Score >= 4
             AND Frequency_Score >= 4
             AND Monetary_Score >= 4
            THEN 'Champions'

        WHEN Recency_Score >= 3
             AND Frequency_Score >= 4
             AND Monetary_Score >= 3
            THEN 'Loyal Customers'

        WHEN Recency_Score >= 4
             AND Frequency_Score >= 2
             AND Monetary_Score >= 2
            THEN 'Potential Loyalists'

        WHEN Recency_Score <= 2
             AND Frequency_Score >= 3
             AND Monetary_Score >= 3
            THEN 'At Risk'

        WHEN Recency_Score <= 2
             AND Frequency_Score <= 2
             AND Monetary_Score <= 2
            THEN 'Lost / Low Value'

        ELSE 'Regular Customers'

    END AS Customer_Segment

FROM rfm_scores

ORDER BY
    RFM_Total_Score DESC,
    Monetary DESC;
"""

rfm_analysis = pd.read_sql_query(
    rfm_query,
    conn
)

display(rfm_analysis)
```

```python
rfm_segment_summary = (
    rfm_analysis
    .groupby("Customer_Segment")
    .agg(
        Customers=("Customer_ID", "count"),
        Revenue=("Monetary", "sum"),
        Average_Recency_Days=("Recency_Days", "mean"),
        Average_Frequency=("Frequency", "mean"),
        Average_Monetary=("Monetary", "mean")
    )
    .reset_index()
)

rfm_segment_summary = rfm_segment_summary.round(2)

display(rfm_segment_summary)
```

```python
# ============================================
# ADVANCED SQL 12: SALES CHANNEL PERFORMANCE
# ============================================

channel_performance_query = """
SELECT
    Channel,

    COUNT(DISTINCT Order_ID) AS Orders,

    COUNT(DISTINCT Customer_ID) AS Customers,

    SUM(Quantity) AS Units_Sold,

    ROUND(SUM(Sales_Amount), 2) AS Revenue,

    ROUND(SUM(Cost_Amount), 2) AS Cost,

    ROUND(SUM(Profit), 2) AS Profit,

    ROUND(AVG(Discount_Percent), 2) AS Average_Discount,

    ROUND(
        SUM(Profit) * 100.0 /
        NULLIF(SUM(Sales_Amount), 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(
        SUM(Sales_Amount) /
        NULLIF(COUNT(DISTINCT Order_ID), 0),
        2
    ) AS Average_Order_Value,

    ROUND(
        SUM(Sales_Amount) /
        NULLIF(SUM(Quantity), 0),
        2
    ) AS Average_Selling_Price

FROM sales

GROUP BY
    Channel

ORDER BY
    Revenue DESC;
"""

channel_performance = pd.read_sql_query(
    channel_performance_query,
    conn
)

display(channel_performance)
```

```python
# ============================================
# ADVANCED SQL 13: CATEGORY PORTFOLIO ANALYSIS
# ============================================

category_portfolio_query = """
WITH category_performance AS (
    SELECT
        p.Category,

        COUNT(DISTINCT p.Product_ID) AS Products,

        COUNT(DISTINCT s.Order_ID) AS Orders,

        COUNT(DISTINCT s.Customer_ID) AS Customers,

        SUM(s.Quantity) AS Units_Sold,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Cost_Amount), 2) AS Cost,

        ROUND(SUM(s.Profit), 2) AS Profit

    FROM sales s

    LEFT JOIN product p
        ON s.Product_ID = p.Product_ID

    GROUP BY
        p.Category
),

ranked_categories AS (
    SELECT
        *,

        RANK() OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Rank,

        RANK() OVER (
            ORDER BY Profit DESC
        ) AS Profit_Rank

    FROM category_performance
),

benchmarks AS (
    SELECT
        AVG(Revenue) AS Average_Category_Revenue,

        AVG(
            Profit * 100.0 /
            NULLIF(Revenue, 0)
        ) AS Average_Category_Margin

    FROM category_performance
)

SELECT
    rc.Category,
    rc.Products,
    rc.Orders,
    rc.Customers,
    rc.Units_Sold,
    rc.Revenue,
    rc.Cost,
    rc.Profit,

    ROUND(
        rc.Profit * 100.0 /
        NULLIF(rc.Revenue, 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(
        rc.Revenue /
        NULLIF(rc.Orders, 0),
        2
    ) AS Average_Order_Value,

    rc.Revenue_Rank,
    rc.Profit_Rank,

    CASE
        WHEN rc.Revenue > b.Average_Category_Revenue
             AND
             (
                 rc.Profit * 100.0 /
                 NULLIF(rc.Revenue, 0)
             ) > b.Average_Category_Margin
            THEN 'High Revenue - High Margin'

        WHEN rc.Revenue > b.Average_Category_Revenue
             AND
             (
                 rc.Profit * 100.0 /
                 NULLIF(rc.Revenue, 0)
             ) < b.Average_Category_Margin
            THEN 'High Revenue - Low Margin'

        WHEN rc.Revenue < b.Average_Category_Revenue
             AND
             (
                 rc.Profit * 100.0 /
                 NULLIF(rc.Revenue, 0)
             ) > b.Average_Category_Margin
            THEN 'Low Revenue - High Margin'

        ELSE 'Low Revenue - Low Margin'

    END AS Portfolio_Segment

FROM ranked_categories rc

CROSS JOIN benchmarks b

ORDER BY
    rc.Revenue_Rank;
"""

category_portfolio = pd.read_sql_query(
    category_portfolio_query,
    conn
)

display(category_portfolio)
```

```python
# ============================================
# ADVANCED SQL 14: REGION × CATEGORY ANALYSIS
# ============================================

region_category_query = """
WITH region_category AS (
    SELECT
        s.Region_ID,
        r.Region,
        p.Category,

        COUNT(DISTINCT s.Order_ID) AS Orders,

        COUNT(DISTINCT s.Customer_ID) AS Customers,

        SUM(s.Quantity) AS Units_Sold,

        ROUND(SUM(s.Sales_Amount), 2) AS Revenue,

        ROUND(SUM(s.Cost_Amount), 2) AS Cost,

        ROUND(SUM(s.Profit), 2) AS Profit

    FROM sales s

    LEFT JOIN region r
        ON s.Region_ID = r.Region_ID

    LEFT JOIN product p
        ON s.Product_ID = p.Product_ID

    GROUP BY
        s.Region_ID,
        r.Region,
        p.Category
),

ranked_matrix AS (
    SELECT
        *,

        RANK() OVER (
            PARTITION BY Region_ID
            ORDER BY Revenue DESC
        ) AS Category_Revenue_Rank_In_Region,

        RANK() OVER (
            PARTITION BY Category
            ORDER BY Revenue DESC
        ) AS Region_Revenue_Rank_In_Category

    FROM region_category
)

SELECT
    Region_ID,
    Region,
    Category,

    Orders,
    Customers,
    Units_Sold,

    Revenue,
    Cost,
    Profit,

    ROUND(
        Profit * 100.0 /
        NULLIF(Revenue, 0),
        2
    ) AS Profit_Margin_Percent,

    ROUND(
        Revenue /
        NULLIF(Orders, 0),
        2
    ) AS Average_Order_Value,

    Category_Revenue_Rank_In_Region,
    Region_Revenue_Rank_In_Category

FROM ranked_matrix

ORDER BY
    Region,
    Revenue DESC;
"""

region_category_analysis = pd.read_sql_query(
    region_category_query,
    conn
)

display(region_category_analysis)
```

```python
# ============================================
# ADVANCED SQL 15: PROFITABILITY & LOSS ANALYSIS
# ============================================

loss_transaction_query = """
SELECT
    Order_ID,
    Order_Date,
    Customer_ID,
    Product_ID,
    Region_ID,
    Channel,

    Quantity,
    Unit_Price,
    Discount_Percent,

    ROUND(Sales_Amount, 2) AS Revenue,
    ROUND(Cost_Amount, 2) AS Cost,
    ROUND(Profit, 2) AS Profit,

    CASE
        WHEN Profit < 0 THEN 'Loss-Making'
        WHEN Profit = 0 THEN 'Break-Even'
        ELSE 'Profitable'
    END AS Profitability_Status

FROM sales

ORDER BY
    Profit ASC;
"""

loss_transaction_analysis = pd.read_sql_query(
    loss_transaction_query,
    conn
)

display(
    loss_transaction_analysis.head(20)
)
```
